# Tuples and Tuple Methods

```python
# A tuple in Python is an ordered, immutable collection of elements. 
# It is similar to a list, but once created, its elements cannot be modified.
```

```python
t = () # Empty Tuple
```

```python
type(t)
```

```python
t
```

### Common Tuple Methods

```python
a = (5,3,78,5,"Happy","Python",6.1,3,5)
```

```python
a
```

```python
type(a)
```

#### count(x) , index(x)

```python
a.count(5) # Returns the number of times x appears in the tuple.
```

```python
a.count(3)
```

```python
a.index(5) # Returns the index of the first occurrence of x .
```

```python
a.index('Happy')
```

### Indexing and Slicing

```python
a1 = (10,20,30,40,50)
```

```python
a1 
```

```python
type(a1)
```

```python
# Indexing
a1.index(30)
```

```python
print(a1[3])
```

```python
# Slicing
print(a1[1:3])
```

```python
print(a1[2:4])
```
