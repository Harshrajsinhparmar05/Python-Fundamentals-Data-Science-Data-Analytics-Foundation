# Dictionary and Dictionary Methods

```python
# A dictionary in Python is an unordered, mutable, and key-value pair collection. 
# It allows efficient data retrieval and modification. Dictionaries in Python are ordered as of Python 3.7
```

```python
d = {}
```

```python
type(d)
```

```python
student = {"Name":"Alice", # Dictionary with key-value pairs
     "Age":18,
     "Grade":"A"
    }
```

student

```python
print(student["Name"]) # Using keys
```

```python
print(student.get("Age")) # # Using get() (avoids KeyError if key doesn't exist)
```

```python
print(student.get("City")) 
```
