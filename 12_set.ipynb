## Set Operations

```python
set1 = {5,7,18,45,"CR7","Ramos",93.28,False,"Kroos",10}
set2 = {7,8,"Marcelo",90.45,10,False,"Modric"}
```

```python
set1.union(set2) # Returns a new set with all unique elements from both sets
```

```python
set1.intersection(set2) # Returns a set with elements common to both sets.
```

```python
set1.difference(set2)  # Returns a set with elements in  set1 but not in  set2 .
```

```python
set1.symmetric_difference(set2) # Returns a set with elements in either set1 or set2 but not both
```

```python
st1 = {1,2,3,4}
st2 = {2,3}
```

```python
st1.issuperset(st2) # Returns True if set1 is a superset of set2 
```

```python
st2.issuperset(st1)
```

```python
st2.issubset(st1) # Returns True if set1 is a subset of set2 .
```

```python
st1.issubset(st2)
```
