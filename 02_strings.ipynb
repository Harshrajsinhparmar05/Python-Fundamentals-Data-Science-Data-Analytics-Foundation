# String and string methods

```python
# In this section, we learn about strings and string methods in Python to work with and manipulate text data
```

```python
# What is a String? 
# A string is a sequence of characters enclosed in single (') or double (") quotes.
```

```python
name = "Harshraj"
```

```python
name
```

```python
type(name)
```

```python
name.upper() # Converts to uppercase
```

```python
name.lower() # Converts to lowercase 
```

```python
name.count("a") # Counts occurrences of substring
```

```python
name.count("j") 
```

```python
name.startswith("Ha") # Checks if string starts with value
```

```python
name.startswith("gs")
```

```python
name.endswith("aj") # Checks if string ends with value 
```

```python
name.endswith("nh") 
```

```python
name.isdigit() # Checks if all chars are digits
```

```python
name.isalpha() # Checks if all chars are letters
```

```python
name.isalnum() # Checks if all chars are letters/digits
```

```python
print(len(name)) # print length of string
```

```python
a = " CodersofAhmedabad  " 
```

```python
a.strip() # Removes leading/trailing spaces
```

```python
a.split() # Splits string into a list
```

```python
a.find("A") # Returns index of first occurrence
```

```python
a.count("d") # Counts occurrences of substring
```

```python
words = ["Python", "is", "powerful"] 
```

```python
sentence = " ".join(words)   # Joins list into string
```

```python
print(sentence) # Output: Python is powerful
```

```python
a1 = "Hello World" # Replaces substring 
```

```python
a2 = a1.replace("World", "Python")
```

```python
a2
```
