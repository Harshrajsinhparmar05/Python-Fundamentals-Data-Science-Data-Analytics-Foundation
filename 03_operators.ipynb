# Operators in Python

```python
"""
There is 6 types of operators in python 
1. Arithmetic Operators
2. Comparison Operators
3. Logical Operators
4. Bitwise Operators
5. Assignment Operators
6. Membership & Identity Operators

"""
```

## 1. Arithmetic Operators 

```python
# Arithmetic Operators Used for basic mathematical operations.
```

```python
a = 54
```

```python
b = 6
```

```python
a + b # Addition
```

```python
a - b # Subtraction
```

```python
a * b # Multiplication
```

```python
a / b #  Division
```

```python
a // b # Floor Division
```

```python
a % b # Modulus
```

```python
a ** b # Exponentiation
```

## 2. Comparison Operators

```python
# Compare values and return True and False
```

```python
a1 = 16
```

```python
b1 = 8
```

```python
a1 == b1     # [a == b] a is equals to b if yes then return true or return false 
```

```python
a1 != b1    # [a != b]a is not equals to b
```

```python
a1 > b1     # [a > b]a is greater than b  
```

```python
a1 < b1     # [a < b] a is less than b
```

```python
a1 >= b1    # [a >= b] a is greater than or equals to b
```

```python
a1 <= b1    # [a <= b] a is less than or equals to b
```
