# Loops in Python

```python
# Python has two main loops: for and while
```

## 1. For Loop

```python
# Used to iterate over sequences like lists, tuples, and strings.
```

```python
fruits = ["Apple","Banana","Mango"]

for fruits in fruits:
    print("Yes")
```

### Using range()

```python
for i in range(10): # it is start from zero (0) to n-1 number.
    print(i)
```

```python
for a in range(1,11): # here we delcare starting numbers to n-1 numbers.
    print(a)
```

## 2. While Loop

```python
# Runs as long as a condition is True
```

```python
count = 0
while count < 3:
    print(count)
    count += 1
```

## 3. Loop Control Statements

```python
# break → Exits the loop.
# continue → Skips to the next iteration.
# pass → Does nothing (used as a placeholder).
```

#### break statement

```python
for i in range(5):
    if i==3:
        break # stops the loop at 3
    print(i)
```

```python
for i in range(5):
    if i==3:
        pass # Does nothing (used as a placeholder)

a = "Hello"
```

#### continue statement

```python
for a in range(10):
    if a==4:
        continue # this will skip the value and continue the loop
    print(a)
```
