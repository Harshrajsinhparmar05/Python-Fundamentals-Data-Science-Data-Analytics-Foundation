## Common List Methods

```python
a = ["apple",18,7,"Python",6.3]
```

```python
a
```

#### append(x)

```python
a.append("45")  # Adds an element x to the end of the list
a.append("Python")
```

```python
a
```

#### extend(iterable)

```python
a1 = ["CondaENV",33] #Extends the list by appending all elements from an iterable
a.extend(a1)
```

```python
a
```

#### insert(index, x)

```python
a.insert(4, 5) # Inserts x at the specified  index . here we insert 5 at index 4.
```

```python
a
```

#### remove(x)

```python
a.remove("Python") # Removes the first occurrence of x in the list.
```

```python
a
```

#### pop([index])

```python
a.pop(5) # Removes and returns the element at index (last element if index is not provided)
```

```python
a
```

#### index(x)

```python
a.index(5) # Returns the index of the first occurrence of x .
```

```python
a.index("CondaENV")
```

```python
a.append(5)
```

```python
a
```

#### count(x)

```python
a.count(5) # Returns the number of times x appears in the list.
```

```python
a.count('Python')
```

#### sort()

```python
b = [4,7,6,2,2,4,46,464,687,21] # Sorts the list in ascending order.
```

```python
b.sort()
```

```python
b
```

#### reverse()

```python
b.reverse() #  Reverses the order of the list.
```

```python
b
```

#### copy()

```python
b1 = b.copy() # Returns a shallow copy of the list
```

```python
b1
```

#### clear()

```python
b1.clear() # Removes all elements from the list.
```

```python
b1
```
