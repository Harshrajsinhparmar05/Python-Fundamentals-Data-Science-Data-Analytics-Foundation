# If else statements

```python
'''
In Python, conditional statements ( if , elif , and  else ) are used to control the
flow of a program based on conditions. These are essential in data science for
handling different scenarios in data processing, decision-making, and logic execution.

'''
```

## Basic If statement

```python
# The if statement allows you to execute a block of code only if a condition is True .
```

```python
x = 15                          # • The condition If x > 10 is checked.True , the indented block under it runs.
if x > 10:                      # • If False , nothing happens. 
    print("Yes")  
```

```python
x1 = int(input("Enter the number: "))
if x1 > 5:
    print(x1, "is greater than 5")
```

## if-else Statement

```python
# The else block executes when the if condition is False.
```

```python
x = 4
```

```python
if x>5:                     # if x>5 is true then it prints first message 
    print("Greater")        # otherwise the else block will execute.
else:
    print("Smaller")
```

```python
x = int(input("Enter the number: "))
y = int(input("Enter the number: "))

if x>y:
    print(x, "is greater than ",y)
else:
    print(x, "is smaller than ",y)
```

```python
a = "Harshraj"

if "sh" in a:
    print("Hello")
```

## If-elif-else Statement

```python
# When multiple conditions need to be checked sequentially, use “else if”).
```

```python
marks = 77
```

```python
if marks > 90:                        #  The conditions are checked from top to bottom.
    print("Congratulations")          #  The first True condition executes, and the rest are skipped.
elif marks > 70:
    print("Good")
elif marks >= 33:
    print("Keep learning")
else:
    print("fail")
```

```python
marks = int(input("Enter the marks: "))

if marks >= 90:
    print("Congratulations")
elif marks >= 70:
    print("Good")
elif marks >= 33:
    print("Keep learning")
else:
    print("fail")
```

## Using if-else in Data Science

```python
# Conditional statements are widely used in data science for filtering, cleaning, and decision-making.
```

```python
# Example: Categorizing Data

age = int(input("Enter Your age: "))

if age <18:
    Category = "Minor"
elif age<65:
    Category= "Adult"
else:
    Category= "Seniot Citizen"
print("Category: ", Category)
```
